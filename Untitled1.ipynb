```python
# Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from datetime import datetime, timedelta
import random

# Set random seed for reproducibility
np.random.seed(42)

# Generate date range from 2000 to 2026
start_date = datetime(2000, 1, 1)
end_date = datetime(2026, 12, 31)
dates = pd.date_range(start=start_date, end=end_date, freq='D')

# Base gold price in 2000 (approx ₹4,400 per 10 grams)
base_price = 4400

# Number of days
n_days = len(dates)

# Generate trend component (gradual increase over time)
# Gold has shown significant appreciation over this period
trend = np.linspace(0, 1, n_days) ** 1.5 * 60000  # Exponential-like growth

# Add cyclical component (seasonal patterns)
# Gold often shows patterns based on wedding seasons, festivals
seasonal = 2000 * np.sin(2 * np.pi * np.arange(n_days) / 365) + \
           1000 * np.sin(2 * np.pi * np.arange(n_days) / 180)

# Add major event impacts (2008 crisis, 2020 pandemic, etc.)
event_impacts = np.zeros(n_days)

# 2008 Financial Crisis spike
crisis_start = datetime(2008, 9, 15)
crisis_end = datetime(2009, 6, 30)
crisis_mask = (dates >= crisis_start) & (dates <= crisis_end)
event_impacts[crisis_mask] = np.linspace(0, 8000, sum(crisis_mask))

# 2020 COVID-19 spike
covid_start = datetime(2020, 3, 1)
covid_end = datetime(2020, 8, 31)
covid_mask = (dates >= covid_start) & (dates <= covid_end)
event_impacts[covid_mask] = np.linspace(0, 12000, sum(covid_mask))

# 2023-2024 Geopolitical tensions
geo_start = datetime(2023, 1, 1)
geo_end = datetime(2024, 12, 31)
geo_mask = (dates >= geo_start) & (dates <= geo_end)
event_impacts[geo_mask] = np.linspace(0, 5000, sum(geo_mask))

# Add noise (random daily fluctuations)
noise = np.random.normal(0, 500, n_days)

# Combine all components
gold_prices = base_price + trend + seasonal + event_impacts + noise

# Ensure no negative prices
gold_prices = np.maximum(gold_prices, 1000)

# Create DataFrame
df = pd.DataFrame({
    'Date': dates,
    'Gold_Price_INR_per_10g': np.round(gold_prices, 2)
})

# Add additional useful columns
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
df['Quarter'] = df['Date'].dt.quarter
df['Day_of_Week'] = df['Date'].dt.day_name()

# Calculate daily returns
df['Daily_Return_%'] = df['Gold_Price_INR_per_10g'].pct_change() * 100

# Calculate rolling averages
df['30D_MA'] = df['Gold_Price_INR_per_10g'].rolling(window=30).mean()
df['90D_MA'] = df['Gold_Price_INR_per_10g'].rolling(window=90).mean()

# Display first few rows
print("Dataset Overview:")
print(f"Total records: {len(df)}")
print(f"Date range: {df['Date'].min()} to {df['Date'].max()}")
print("\nFirst 10 rows:")
print(df.head(10))

# Basic statistics
print("\nDataset Statistics:")
print(df['Gold_Price_INR_per_10g'].describe())

# Save to CSV
df.to_csv('indian_gold_prices_2000_2026.csv', index=False)
print("\nDataset saved to 'indian_gold_prices_2000_2026.csv'")

# Visualization
fig, axes = plt.subplots(2, 2, figsize=(15, 10))

# 1. Full time series
axes[0, 0].plot(df['Date'], df['Gold_Price_INR_per_10g'], linewidth=0.5, color='gold')
axes[0, 0].set_title('Indian Gold Price (2000-2026)')
axes[0, 0].set_xlabel('Year')
axes[0, 0].set_ylabel('Price (INR per 10g)')
axes[0, 0].grid(True, alpha=0.3)

# 2. Yearly average prices
yearly_avg = df.groupby('Year')['Gold_Price_INR_per_10g'].mean()
axes[0, 1].bar(yearly_avg.index, yearly_avg.values, color='goldenrod')
axes[0, 1].set_title('Average Yearly Gold Prices')
axes[0, 1].set_xlabel('Year')
axes[0, 1].set_ylabel('Avg Price (INR per 10g)')
axes[0, 1].tick_params(axis='x', rotation=45)

# 3. Monthly pattern (seasonality)
monthly_avg = df.groupby('Month')['Gold_Price_INR_per_10g'].mean()
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 
          'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
axes[1, 0].plot(months, monthly_avg.values, marker='o', color='darkgoldenrod')
axes[1, 0].set_title('Average Monthly Gold Prices')
axes[1, 0].set_xlabel('Month')
axes[1, 0].set_ylabel('Avg Price (INR per 10g)')
axes[1, 0].grid(True, alpha=0.3)

# 4. Distribution of prices
axes[1, 1].hist(df['Gold_Price_INR_per_10g'], bins=50, color='gold', edgecolor='black')
axes[1, 1].set_title('Distribution of Gold Prices')
axes[1, 1].set_xlabel('Price (INR per 10g)')
axes[1, 1].set_ylabel('Frequency')
axes[1, 1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# Additional analysis
print("\nKey Insights:")
print(f"Highest price: ₹{df['Gold_Price_INR_per_10g'].max():,.2f} on {df.loc[df['Gold_Price_INR_per_10g'].idxmax(), 'Date'].strftime('%Y-%m-%d')}")
print(f"Lowest price: ₹{df['Gold_Price_INR_per_10g'].min():,.2f} on {df.loc[df['Gold_Price_INR_per_10g'].idxmin(), 'Date'].strftime('%Y-%m-%d')}")
print(f"Average price: ₹{df['Gold_Price_INR_per_10g'].mean():,.2f}")
print(f"Total appreciation: {(df['Gold_Price_INR_per_10g'].iloc[-1] / df['Gold_Price_INR_per_10g'].iloc[0] - 1) * 100:.2f}%")
```

```python
# Import required libraries
import yfinance as yf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from datetime import datetime
import warnings
warnings.filterwarnings('ignore')

# Set date range from 2000 to 2026
start_date = '2000-01-01'
end_date = '2026-12-31'

print(f"Fetching real gold price data from {start_date} to {end_date}...")

# Fetch gold price data from yfinance
# GC=F is the gold futures contract (continuous contract)
gold_data = yf.download('GC=F', start=start_date, end=end_date, progress=False)

if gold_data.empty:
    print("Note: yfinance may have limited historical data. Using available data...")
    # Try alternative method - get maximum available data
    gold_ticker = yf.Ticker('GC=F')
    gold_data = gold_ticker.history(period='max')
    
# Display the data structure
print("\nData retrieved successfully!")
print(f"Date range: {gold_data.index.min()} to {gold_data.index.max()}")
print(f"Total trading days: {len(gold_data)}")
print("\nFirst 5 rows:")
print(gold_data.head())

# The price in yfinance is in USD per troy ounce
# Convert to Indian price (INR per 10 grams)

# 1 troy ounce = 31.1035 grams
# So price per 10 grams in USD = (USD per ounce) * (10/31.1035)

# We need USD/INR exchange rate data
print("\nFetching USD/INR exchange rate data...")
usdinr_data = yf.download('INR=X', start=start_date, end=end_date, progress=False)

if usdinr_data.empty:
    print("Using alternative method for USD/INR...")
    usdinr_ticker = yf.Ticker('INR=X')
    usdinr_data = usdinr_ticker.history(period='max')

print(f"Exchange rate data range: {usdinr_data.index.min()} to {usdinr_data.index.max()}")

# Merge the datasets
combined_data = pd.DataFrame(index=gold_data.index)
combined_data['Gold_USD_per_Oz'] = gold_data['Close']
combined_data['USDINR'] = usdinr_data['Close']

# Forward fill missing exchange rate values (for weekends/holidays)
combined_data['USDINR'] = combined_data['USDINR'].fillna(method='ffill')

# Drop rows with missing values
combined_data = combined_data.dropna()

# Convert to Indian price (INR per 10 grams)
# Formula: (USD/oz) * (10/31.1035) * USDINR
combined_data['Gold_INR_per_10g'] = combined_data['Gold_USD_per_Oz'] * (10/31.1035) * combined_data['USDINR']

# Round to 2 decimal places
combined_data['Gold_INR_per_10g'] = combined_data['Gold_INR_per_10g'].round(2)

# Create final DataFrame
df = pd.DataFrame({
    'Date': combined_data.index,
    'Gold_Price_INR_per_10g': combined_data['Gold_INR_per_10g'],
    'Gold_Price_USD_per_Oz': combined_data['Gold_USD_per_Oz'].round(2),
    'USD_INR_Rate': combined_data['USDINR'].round(2)
})

# Add useful columns
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
df['Quarter'] = df['Date'].dt.quarter
df['Day'] = df['Date'].dt.day
df['Day_of_Week'] = df['Date'].dt.day_name()
df['Week_Number'] = df['Date'].dt.isocalendar().week

# Calculate daily returns
df['Daily_Return_%'] = df['Gold_Price_INR_per_10g'].pct_change() * 100

# Calculate moving averages
df['30D_MA'] = df['Gold_Price_INR_per_10g'].rolling(window=30).mean()
df['90D_MA'] = df['Gold_Price_INR_per_10g'].rolling(window=90).mean()
df['200D_MA'] = df['Gold_Price_INR_per_10g'].rolling(window=200).mean()

# Calculate yearly statistics
df['Yearly_High'] = df.groupby('Year')['Gold_Price_INR_per_10g'].transform('max')
df['Yearly_Low'] = df.groupby('Year')['Gold_Price_INR_per_10g'].transform('min')

print("\n" + "="*60)
print("FINAL DATASET SUMMARY")
print("="*60)
print(f"Date range: {df['Date'].min()} to {df['Date'].max()}")
print(f"Total records: {len(df)}")
print(f"Years covered: {df['Year'].nunique()}")
print(f"Average gold price: ₹{df['Gold_Price_INR_per_10g'].mean():,.2f}")
print(f"Highest price: ₹{df['Gold_Price_INR_per_10g'].max():,.2f} on {df.loc[df['Gold_Price_INR_per_10g'].idxmax(), 'Date'].strftime('%Y-%m-%d')}")
print(f"Lowest price: ₹{df['Gold_Price_INR_per_10g'].min():,.2f} on {df.loc[df['Gold_Price_INR_per_10g'].idxmin(), 'Date'].strftime('%Y-%m-%d')}")

print("\nFirst 10 rows of the dataset:")
print(df.head(10).to_string())

# Save to CSV
csv_filename = 'indian_gold_prices_2000_2026_real.csv'
df.to_csv(csv_filename, index=False)
print(f"\n✅ Dataset saved to '{csv_filename}'")

# Also save yearly summary
yearly_summary = df.groupby('Year').agg({
    'Gold_Price_INR_per_10g': ['min', 'max', 'mean', 'std'],
    'Daily_Return_%': 'mean'
}).round(2)

yearly_summary.columns = ['Min_Price', 'Max_Price', 'Avg_Price', 'Std_Dev', 'Avg_Daily_Return_%']
yearly_summary = yearly_summary.reset_index()

yearly_filename = 'indian_gold_prices_yearly_summary.csv'
yearly_summary.to_csv(yearly_filename, index=False)
print(f"✅ Yearly summary saved to '{yearly_filename}'")
```

```python
# Alternative approach using known historical yearly data with interpolation
print("\n" + "="*60)
print("ALTERNATIVE: HYBRID APPROACH WITH INTERPOLATION")
print("="*60)

# Use known historical yearly averages [citation:4][citation:8]
historical_yearly = {
    2000: 4400, 2001: 4300, 2002: 4990, 2003: 5600, 2004: 5850,
    2005: 7000, 2006: 8490, 2007: 10800, 2008: 12500, 2009: 14500,
    2010: 18500, 2011: 26400, 2012: 31050, 2013: 29600, 2014: 28006.50,
    2015: 26343.50, 2016: 28623.50, 2017: 29667.50, 2018: 31438,
    2019: 35220, 2020: 48651, 2021: 48720, 2022: 52670,
    2023: 65330, 2024: 77913, 2025: 105000, 2026: 157400
}

# Generate daily dates
all_dates = pd.date_range(start='2000-01-01', end='2026-12-31', freq='D')

# Create base DataFrame with yearly averages
yearly_df = pd.DataFrame(list(historical_yearly.items()), columns=['Year', 'Avg_Price'])
yearly_df['Date'] = pd.to_datetime(yearly_df['Year'].astype(str) + '-07-01')  # Mid-year point

# Interpolate to daily
daily_prices = []
for date in all_dates:
    year = date.year
    if year in historical_yearly:
        # Add seasonal variation and noise
        base_price = historical_yearly[year]
        
        # Add seasonal pattern (higher during festival seasons)
        month_factor = 1 + 0.03 * np.sin(2 * np.pi * (date.timetuple().tm_yday - 15) / 365)
        
        # Add daily volatility
        daily_volatility = np.random.normal(0, base_price * 0.005)  # 0.5% daily volatility
        
        # Add trend within year
        if year < 2026:
            next_year_price = historical_yearly[year + 1]
            year_progress = (date.timetuple().tm_yday - 1) / 365
            trend_component = (next_year_price - base_price) * year_progress
        else:
            trend_component = 0
        
        price = (base_price + trend_component) * month_factor + daily_volatility
        daily_prices.append({
            'Date': date,
            'Gold_Price_INR_per_10g': round(price, 2)
        })

# Create hybrid dataset
hybrid_df = pd.DataFrame(daily_prices)
hybrid_df.to_csv('indian_gold_prices_hybrid.csv', index=False)
print(f"\n✅ Hybrid dataset saved to 'indian_gold_prices_hybrid.csv'")
print(f"Records created: {len(hybrid_df)}")
print(f"Date range: {hybrid_df['Date'].min()} to {hybrid_df['Date'].max()}")
```

```python
print("\n" + "="*70)
print("INVESTMENT ANALYSIS BY CARAT")
print("="*70)

# Calculate investment metrics
initial_price_24k = df['Gold_24K_INR_per_10g'].iloc[0]
current_price_24k = df['Gold_24K_INR_per_10g'].iloc[-1]
years = (df['Date'].iloc[-1] - df['Date'].iloc[0]).days / 365.25

investment_data = []
for carat in ['24K', '22K', '18K']:
    col = f'Gold_{carat}_INR_per_10g'
    initial = df[col].iloc[0]
    current = df[col].iloc[-1]
    total_return = ((current / initial) - 1) * 100
    cagr = (pow(current / initial, 1/years) - 1) * 100
    
    investment_data.append({
        'Carat': carat,
        'Initial (2000)': f"₹{initial:,.2f}",
        'Current (2026)': f"₹{current:,.2f}",
        'Total Return': f"{total_return:.2f}%",
        'CAGR': f"{cagr:.2f}%",
        'Value Growth': f"₹{(current - initial):,.2f}"
    })

investment_df = pd.DataFrame(investment_data)
print(investment_df.to_string(index=False))

# Calculate best and worst years for each carat
print("\n" + "="*70)
print("BEST AND WORST YEARS BY CARAT")
print("="*70)

yearly_returns = df.groupby('Year').agg({
    'Gold_24K_INR_per_10g': lambda x: (x.iloc[-1] / x.iloc[0] - 1) * 100,
    'Gold_22K_INR_per_10g': lambda x: (x.iloc[-1] / x.iloc[0] - 1) * 100,
    'Gold_18K_INR_per_10g': lambda x: (x.iloc[-1] / x.iloc[0] - 1) * 100
}).round(2)

for carat in ['24K', '22K', '18K']:
    col = f'Gold_{carat}_INR_per_10g'
    best_year = yearly_returns[col].idxmax()
    worst_year = yearly_returns[col].idxmin()
    print(f"\n{carat} Gold:")
    print(f"  Best Year: {int(best_year)} with {yearly_returns.loc[best_year, col]:.2f}% return")
    print(f"  Worst Year: {int(worst_year)} with {yearly_returns.loc[worst_year, col]:.2f}% return")

# Create carat conversion calculator function
print("\n" + "="*70)
print("CARAT CONVERSION CALCULATOR")
print("="*70)

def convert_carat(price_24k_per_10g, target_carat):
    """
    Convert 24K gold price to other caratages
    """
    purity_factors = {
        '24K': 1.0,
        '22K': 22/24,
        '21K': 21/24,
        '18K': 18/24,
        '14K': 14/24,
        '12K': 12/24,
        '10K': 10/24,
        '9K': 9/24
    }
    
    if target_carat in purity_factors:
        return price_24k_per_10g * purity_factors[target_carat]
    else:
        return None

# Example usage with latest price
latest_24k = df['Gold_24K_INR_per_10g'].iloc[-1]
print(f"\nBased on latest 24K price: ₹{latest_24k:,.2f} per 10g")
print("\nEquivalent prices:")

for carat in ['24K', '22K', '21K', '18K', '14K', '12K', '10K', '9K']:
    converted = convert_carat(latest_24k, carat)
    print(f"  {carat}: ₹{converted:,.2f} per 10g (₹{converted/10:,.2f} per gram)")
```

```python
import yfinance as yf
import pandas as pd

start = "2004-01-01"
end = "2026-12-31"

# Gold price (USD per ounce)
gold = yf.download("GC=F", start=start, end=end)

# USD to INR
usd_inr = yf.download("USDINR=X", start=start, end=end)

# Merge datasets
data = pd.DataFrame()
data["gold_usd_oz"] = gold["Close"]
data["usd_inr"] = usd_inr["Close"]

# Forward fill missing values
data = data.fillna(method="ffill")

# Convert ounce -> gram
ounce_to_gram = 31.1035

# Convert to INR per gram (24K)
data["price_24k"] = (data["gold_usd_oz"] * data["usd_inr"]) / ounce_to_gram

# Calculate other purity levels
data["price_22k"] = data["price_24k"] * 0.916
data["price_18k"] = data["price_24k"] * 0.75

# Round values
data["price_24k"] = data["price_24k"].round(2)
data["price_22k"] = data["price_22k"].round(2)
data["price_18k"] = data["price_18k"].round(2)

# Keep useful columns
dataset = data[["price_24k","price_22k","price_18k"]]

# Save dataset
dataset.to_csv("gold_price_india_2000_2025.csv")

print(dataset.head())
print("Dataset saved: gold_price_india_2000_2025.csv")
```
